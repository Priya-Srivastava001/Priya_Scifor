```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data= pd.read_csv("/content/data.csv")
data
```

### Checking for Missing Values

```python
data.isnull().sum()
```

### Handling Missing Values

```python
# Handling missing values by filling with mean for numeric columns and mode for categorical columns
for column in data.columns:
    if data[column].dtype == 'object':
        data[column].fillna(data[column].mode()[0], inplace=True)
    else:
        data[column].fillna(data[column].mean(), inplace=True)

# Verify no missing values
print(data.isnull().sum().sum())
```

### Bar Chart

```python
# Example Bar Chart: Distribution of Sale Condition
plt.figure(figsize=(10, 6))
sns.countplot(x='SaleCondition', data=data)
plt.title('Distribution of Sale Condition')
plt.xlabel('Sale Condition')
plt.ylabel('Count')
plt.show()
```

### Pie Chart

```python
# Example Pie Chart: Distribution of Sale Condition
sale_condition_counts = data['SaleCondition'].value_counts()
plt.figure(figsize=(8, 8))
plt.pie(sale_condition_counts, labels=sale_condition_counts.index, autopct='%1.1f%%')
plt.title('Distribution of Sale Condition')
plt.show()
```

### Line Chart

```python
# Example Line Chart: Average Sale Price Over Months Sold
monthly_sales = data.groupby('MoSold')['SalePrice'].mean().reset_index()
plt.figure(figsize=(10, 6))
sns.lineplot(x='MoSold', y='SalePrice', data=monthly_sales, marker='o')
plt.title('Average Sale Price Over Months Sold')
plt.xlabel('Month Sold')
plt.ylabel('Average Sale Price')
plt.show()
```

### Histogram

```python
# Example Histogram: Distribution of Sale Prices
plt.figure(figsize=(10, 6))
sns.histplot(data['SalePrice'], bins=30, kde=True)
plt.title('Distribution of Sale Prices')
plt.xlabel('Sale Price')
plt.ylabel('Frequency')
plt.show()
```

### Qstn:- What's the use of pandas, numpy, seaborn, matplotlib, OS library in python.
### Ans: **Pandas** is a powerful library for data manipulation and analysis. It provides data structures such as DataFrames and Series, which make it easy to handle and analyze structured data. Key features include:
- Reading and writing data from various file formats (CSV, Excel, SQL, etc.).
- Data cleaning and preprocessing.
- Merging and joining datasets.
- Data aggregation and grouping.

### **NumPy** is a library for numerical computing in Python. It provides support for large, multi-dimensional arrays and matrices, along with a collection of mathematical functions to operate on these arrays efficiently. Key features include:
- Efficient array operations.
- Mathematical and statistical functions.
- Linear algebra and random number generation.

### **Seaborn** is a statistical data visualization library based on Matplotlib. It provides a high-level interface for drawing attractive and informative statistical graphics. Key features include:
- Visualization of univariate and bivariate data.
- Easy-to-use functions for complex plots (e.g., violin plots, box plots).
- Integration with Pandas DataFrames

### **Matplotlib** is a comprehensive library for creating static, animated, and interactive visualizations in Python. It is widely used for plotting a wide variety of graphs and charts. Key features include:
- Line plots, scatter plots, bar charts, histograms, etc.
- Customizable plots with fine control over the appearance.
- Support for interactive plots with tools like Jupyter notebooks.

### OS is a standard library in Python that provides functions for interacting with the operating system. It allows you to perform tasks such as file and directory manipulation, environment variable access, and process management. Key features include:
- Working with file paths.
- Reading and writing files.
- Executing shell commands.
